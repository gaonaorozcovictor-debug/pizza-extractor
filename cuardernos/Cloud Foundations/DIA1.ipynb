¿Qué es un CIDR-Block y cómo se calculan las IPs disponibles?
Se usa para definir el rango de ips disponibles

Para un /24: 2^(32-24) = 2^8 = 256 direcciones en total, así sería para calcular las ips

¿Cuál es el CIDR más grande y más pequeño que se permite en AWS?
● Más grande (mayor cantidad de IPs): /16 (65.536 direcciones IP totales)
● Más pequeño (menor cantidad de IPs): /28 (16 direcciones IP totales)

Es el encargado de que podamos tener conexión entre la VPC e internet y así te puedas conectar
